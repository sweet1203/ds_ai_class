1. 데이터 불러오기

```python
import seaborn as sns
tips_df = sns.load_dataset('tips')
tips_df.head()
```

2. 속성별 결측치 숫자 확인하기

```python
tips_df.isna().sum()
```

3. 통계 정보 확인하기

```python
tips_df.describe()
```

4. 범주형 속성을 숫자형으로 변환하기

```python
import pandas as pd
dummyDf = pd.get_dummies(tips_df)
dummyDf
```

5. 사분위수로 이상치 확인하여 처리하기

```python
q1 = dummyDf['total_bill'].quantile(0.25) # 1사분위수
q3 =  dummyDf['total_bill'].quantile(0.75) # 3사분위수
IQR = q3 - q1 # 사분위수 범위
print('Q1: %.2f, Q3:%.2f, IQR: %.2f'%(q1, q3, IQR))
lower = q1 - 1.5*IQR # 안울타리값
upper = q3 + 1.5*IQR # 밖울타리값
print('lower:%.2f, upper:%.2f'%(lower, upper))
dummyDf[(dummyDf['total_bill'] < lower ) | (dummyDf['total_bill'] > upper)]
```

6. 히트맵으로 속성 간의 관계 파악하여 핵심 속성 찾기

```python
%matplotlib inline
import matplotlib.pyplot as plt

tipData = dummyDf.corr(method='pearson')
plt.figure(figsize=(10,10))
sns.heatmap(data=tipData, annot=True, fmt='.2f', linewidth=.5, cmap='RdYlGn_r');
#상관관계가 저장된 data_f의 데이터들의 상관관계를 heatmap을 이용해 출력
plt.show();
```
