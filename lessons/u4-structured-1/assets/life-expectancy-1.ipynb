## [3] 데이터 전처리
### ① 결측치 확인하기

```python
df.shape    # (데이터 개수, 속성 개수)
```

```python
df.isna().sum()    # 속성별 결측치 개수
```

### ② 결측치 제거하기

```python
df = df.dropna()    # 결측치 데이터 삭제
df.isna().sum()
```

```python
df.shape    # (사본에 더함) 결측치를 지운 뒤 남은 데이터 개수
```

`Status` 속성을 범주형 데이터로 바꾸고, 문자로 된 데이터는 학습할 수 없으므로 개발도상국(Developing)은 0, 선진국(Developed)은 1로 바꿔요.

> 실행하면 분홍색 상자에 `FutureWarning`이 뜰 수 있어요. **오류가 아니라 안내(경고)**라서 결과는 그대로 맞아요. 다음 셀로 넘어가면 돼요.

```python
df['Status'] = df['Status'].astype(object)
df = df.replace({'Developing': 0, 'Developed': 1})    # 개발도상국은 0, 선진국은 1
df['Status'] = df['Status'].astype(int)    # (사본에 더함) 판다스 버전과 상관없이 0과 1을 숫자로
df.head()
```

## [4] 데이터 시각화 및 분석
히트맵(③)과 pairplot(④) 셀은 그리는 데 10~20초쯤 걸려요. 기다리는 동안 수업 화면의 상관관계 탐색기를 살펴봐요.

### ① 기대 수명에 대한 히스토그램

```python
df['Life expectancy'].hist()    # 종속 변수에 대한 히스토그램
```

### ② 소아마비 예방 접종률에 대한 히스토그램
`Polio`는 1세 아동의 **소아마비 예방 접종률(%)**이에요.

```python
import seaborn as sns    # 밀도 추정 곡선과 히스토그램 보여 주기
sns.displot(x = df['Polio'], kde = True)    # 소아마비 변수로 분포도 표시
```

### ③ 히트맵으로 속성 간 상관관계 파악하기

```python
plt.figure(figsize = (15, 12))    # 표 사이즈 설정
sns.heatmap(df.corr(numeric_only = True), linewidth = 0.1, vmax = 1,
            cmap = plt.cm.YlOrRd, linecolor = 'white', annot = True)
```

### ④ 산점도로 특정 속성 간 상관관계 파악하기

```python
sns.pairplot(data = df[['Measles', 'GDP', 'Status', 'Life expectancy']],
             hue = 'Status', palette = 'bright');    # 상관관계를 파악하기
```

### ⑤ 바이올린 그래프 시각화하기

```python
import plotly.express as px
fig = px.violin(df, x = 'Status', y = 'Life expectancy', color = 'Status', box = True)
fig.show()
```

### 해 보기 · 데이터 시각화
다양한 시각화 방법 중 한 개를 선택하여 시각화해 보고, 기대 수명에 영향을 미치는 속성을 적어 보세요.
아래 셀의 속성 이름을 바꿔 가며 산점도를 그려 볼 수 있어요. (사본에 더한 셀)

```python
col = 'Schooling'    # 'Adult Mortality', 'HIV/AIDS', ' BMI', 'GDP' 등으로 바꿔 보기
plt.scatter(df[col], df['Life expectancy'], s = 8)
plt.xlabel(col)
plt.ylabel('Life expectancy')
plt.show()
print('상관계수:', round(df[col].corr(df['Life expectancy']), 2))
```

### 더 알아보기 · 결측치를 지우면 사라지는 나라 (사본에 더한 셀 · 선택)
`dropna()`는 결측치가 **하나라도** 있는 행을 지워요. 어떤 나라가 통째로 사라졌는지 확인해 봐요.

```python
raw = pd.read_csv('https://raw.githubusercontent.com/sweet1203/ds_ai_class/main/lessons/u4-structured-1/assets/life-expectancy-data.csv')    # 결측치를 지우기 전의 원래 데이터
gone = sorted(set(raw['Country']) - set(df['Country']))
print('사라진 나라 수:', len(gone))
print(gone)
raw[raw['Country'] == 'Republic of Korea'].isna().sum()    # 우리나라 데이터의 결측치
```
