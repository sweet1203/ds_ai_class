### 2 데이터셋 불러오기
한글이 포함된 파일을 처리하기 위해 cp949 인코딩 방식을 추가해요.

````python
import pandas as pd    # pandas 모듈을 불러와 별칭을 pd로 지정
df1 = pd.read_csv('https://raw.githubusercontent.com/sweet1203/ds_ai_class/main/lessons/u2-preprocess-bias/assets/Iris1.csv', encoding = 'cp949') # 데이터프레임으로 변환
df2 = pd.read_csv('https://raw.githubusercontent.com/sweet1203/ds_ai_class/main/lessons/u2-preprocess-bias/assets/Iris2.csv', encoding = 'cp949')
df1.head()    # df1의 앞 5개 데이터 출력
````

같은 방법으로 df2 데이터도 출력해 보자.

````python
df2.head()
````

### 3 두 개의 데이터 파일 합치기
df1 내용 뒤에 df2 내용이 붙도록 합하여 df 변수에 저장한다.

````python
df = pd.concat([df1, df2]) # df1과 df2를 위아래 방향으로 합치기
df
````

### 더 알아보기
합친 데이터의 크기(행, 열)와 종류별 개수를 확인해 보자. 빈칸(NaN)이 있는 행도 찾아보자.

````python
print(df.shape)   # (행의 수, 열의 수)
df['종류'].value_counts()
````

````python
df[df.isna().any(axis = 1)]   # 비어 있는 값이 하나라도 있는 행
````
